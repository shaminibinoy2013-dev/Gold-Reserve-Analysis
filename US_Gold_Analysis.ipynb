## Stage 1 – Problem Definition and Dataset Selection




```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

* Load Dataset

```python
url="https://raw.githubusercontent.com/shaminibinoy2013-dev/Gold-Reserve-Analysis/refs/heads/main/U.S.Gold_Reverse_Analysis.csv"
df=pd.read_csv(url)
df.head()
```

#  Understand data


*Initial EDA (head, info, describe, shape, null checks, duplicate check)

```python
df.info()
```

```python
df.describe()
```

```python
df.describe(include='all')
```

```python
df.columns
```

*Data Cleaning

```python
# Remove duplicates
df = df.drop_duplicates()
```

```python
# Handle missing values
df = df.ffill()
```

Forward fill is used to replace missing values using previous valid data.

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

3. Rename Columns

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
```

```python
df.columns
```

Drop unwanted Column

```python
if 'source_line_number' in df.columns:
    df = df.drop(columns=['source_line_number'])
```

4. Outlier Check

```python
plt.figure(figsize=(2,2))
sns.boxplot(x=df['fine_troy_ounces'])
plt.title("Outlier Detection")
plt.show()
```

Interpretation
* The box (blue area) = where most of your data lies (interquartile range).
* The line inside the box = median value.
* The whiskers = spread of typical values.
* The single point on the right = an outlier.

* Skewness Check

```python
print("Skewness:", df['fine_troy_ounces'].skew())
```

# Stage 3 * EDA and Visualizations

# 🔸 Countplot
Understand one column at a time

```python
plt.figure(figsize=(4,5))
sns.countplot(x='facility_description', data=df)
plt.title('Count of Gold by Facility')
plt.xticks(rotation=45)
plt.show()
```

# Interpretation:
* Most gold is stored in a few key facilities

* Some facilities have very low counts → less importance

# 🔸 Histogram


```python
plt.figure(figsize=(4,4))
sns.histplot(df['fine_troy_ounces'], bins=30, kde=True)
plt.title('Distribution of Gold Quantity')
plt.show()
```

**Interpretation**:
* Data is right-skewed → few locations hold very high gold
* Majority of values are smaller

# 🔸 Boxplot

```python
plt.figure(figsize=(4,4))
sns.boxplot(x=df['fine_troy_ounces'])
plt.title('Outliers in Gold Quantity')
plt.show()
```

**Interpretation**:
* Presence of outliers → some facilities hold extremely large reserves
* Important for business: high concentration risk

## 2. Bivariate Analysis (Two Variables)
Relationship between two columns

# 🔸 Scatterplot

```python
sns.scatterplot(x='fine_troy_ounces', y='book_value', data=df)
plt.title('Gold Quantity vs Book Value')
plt.figure(figsize=(4,4))
plt.show()
```

**Interpretation:**
* Strong positive relationship → more gold = higher value
* Business insight: valuation depends directly on quantity

# 🔸 Barplot

```python
plt.figure(figsize=(4,4))
sns.barplot(x='facility_description', y='fine_troy_ounces', data=df)
plt.xticks(rotation=45)

plt.title('Gold by Facility')
plt.show()
```

**Interpretation:**
* Identifies top contributing facilities
* Helps in resource allocation decisions

# Correlation Heatmap

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

**Interpretation:**
* High correlation between Fine Troy Ounces & Book Value
* Useful for prediction models

# 3. Multivariate Analysis (3+ Variables)

 Deep insights using multiple features

# 🔸 Pairplot

```python
sns.pairplot(df[['fine_troy_ounces', 'book_value', 'fiscal_year']])
plt.show()
```

Shows relationships between all numerical variables

Helps detect patterns and clusters

# 🔸 Grouped Analysis

```python
df.groupby('facility_description')['fine_troy_ounces'].mean().sort_values().plot(kind='bar')
plt.title('Average Gold per Facility')
plt.show()
```

Shows average distribution

Helps compare performance of facilities

# 🔸 Pivot Table

```python
pivot = df.pivot_table(values='fine_troy_ounces',
                       index='facility_description',
                       columns='fiscal_year',
                       aggfunc='sum')

pivot.plot(figsize=(10,5))
plt.title('Gold Trend by Facility Over Years')
plt.show()
```

Tracks growth or decline over time

Business insight: stability of reserves

# SUMMARY:
* The dataset contains approximately 31.64 billion Fine Troy Ounces of gold with a total Book Value of around $1.33 trillion, highlighting its financial significance.
* Fort Knox holds the largest share (about 56%), making it the primary gold storage facility in the U.S.
* West Point (~21%) and Denver (~17%) are the next major contributors, showing that most gold reserves are concentrated in a few key locations.
* Over 90% of total gold reserves are stored in just three major facilities, ensuring centralized and secure storage.
* Year-wise analysis shows that gold reserves remain highly stable with no major fluctuations, indicating consistent reserve management.
* Overall, the analysis reveals that U.S. gold reserves are securely concentrated, financially stable, and consistently maintained over time.

# Key Insight:
* Fort Knox holds ~56% of total gold reserves, making it the primary storage location.
* West Point (~21%) and Denver (~17%), together with Fort Knox, account for over 93% of total reserves, showing high concentration.
* The Federal Reserve Bank (NY Vault) contributes ~5%, while all other locations together hold less than 2%.
* Gold reserves show almost 0% fluctuation over the years, indicating strong stability and no major changes in holdings.
* Overall, the data shows that gold reserves are highly centralized, stable, and securely maintained across a few major facilities.
